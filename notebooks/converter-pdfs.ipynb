## Conversão — PDFs em Markdown

Esta célula processa **todos** os PDFs entregues (Opção A, Opção B ou as duas). Para cada livro ela mostra o resultado: `OK` ou um aviso de **PDF escaneado** (quando o arquivo é uma imagem das páginas e não tem texto selecionável — a solução está no Guia 02 do repositório).

A conversão de livros grandes pode levar de alguns segundos a alguns minutos por arquivo. É normal.

```python
# CELULA 4 — Converter todos os PDFs para Markdown (incremental)
import hashlib, json, pathlib, re, unicodedata

# junta o que veio da Opção A e da Opção B (o que não foi usado fica vazio)
lista_pdfs = globals().get('pdfs_subidos', []) + globals().get('pdfs_drive', [])

if not lista_pdfs:
    raise SystemExit("Nenhum PDF entregue. Use a Opção A ou a Opção B antes de executar esta célula.")

def nome_seguro(texto):
    """Remove acentos e espaços para criar nomes de arquivo compatíveis com o GitHub."""
    texto = unicodedata.normalize('NFKD', texto).encode('ascii', 'ignore').decode()
    return re.sub(r'[^a-zA-Z0-9]+', '-', texto).strip('-').lower()

def tema_do_pdf(caminho):
    """Se o PDF veio do Drive em uma subpasta de tema, usa o nome da subpasta."""
    pasta = pathlib.Path(caminho).parent
    raiz = pasta.name  # nome imediato da pasta que contém o PDF
    if raiz and raiz != 'MyDrive':
        return nome_seguro(raiz)
    return 'sem-tema'

def texto_do_pdf(caminho):
    """Extrai o texto pagina a pagina (mais robusto que o modo de uma passada)."""
    try:
        paginas = pymupdf4llm.to_markdown(str(caminho), page_chunks=True)
        partes = [p.get('text', '') if isinstance(p, dict) else str(p) for p in paginas]
        return '\n\n'.join(partes)
    except TypeError:
        # versao antiga da biblioteca: sem page_chunks
        return pymupdf4llm.to_markdown(str(caminho))

saida = pathlib.Path('/content/markdown')
saida.mkdir(exist_ok=True)
MANIFESTO = saida / 'manifesto.json'
manifesto = json.loads(MANIFESTO.read_text(encoding='utf-8')) if MANIFESTO.exists() else {}

relatorio = []
for caminho in lista_pdfs:
    caminho = pathlib.Path(caminho)
    nome = caminho.stem
    destino = saida / (nome_seguro(nome) + '.md')
    sha = hashlib.sha256(caminho.read_bytes()).hexdigest()
    tema = tema_do_pdf(caminho)

    if manifesto.get(nome, {}).get('sha256') == sha and destino.exists():
        print(f"Pulando (já convertido e inalterado): {nome}")
        relatorio.append((nome, 'JA CONVERTIDO', len(destino.read_text(encoding='utf-8')), tema))
        continue

    print(f"Convertendo: {nome} ...", end=' ')
    try:
        md_texto = texto_do_pdf(caminho)
    except Exception as erro:
        relatorio.append((nome, 'ERRO', 0, tema))
        print(f"ERRO ({erro})")
        continue
    if len(md_texto.strip()) < 300:
        relatorio.append((nome, 'SEM TEXTO (provavelmente escaneado)', 0, tema))
        manifesto[nome] = {'sha256': sha, 'status': 'sem-texto', 'tema': tema}
        print("sem texto — provavelmente PDF escaneado (veja o Guia 02)")
        continue

    destino.write_text(md_texto, encoding='utf-8')
    manifesto[nome] = {'sha256': sha, 'status': 'ok', 'tema': tema,
                       'arquivo': destino.name, 'caracteres': len(md_texto)}
    relatorio.append((nome, 'OK', len(md_texto), tema))
    print(f"OK ({len(md_texto):,} caracteres).".replace(',', '.'))

MANIFESTO.write_text(json.dumps(manifesto, ensure_ascii=False, indent=2), encoding='utf-8')

print()
print("=== RESUMO DA CONVERSÃO ===")
for nome, status, tamanho, tema in relatorio:
    print(f"{status:>35} | {tema:<14} | {nome}")
```
